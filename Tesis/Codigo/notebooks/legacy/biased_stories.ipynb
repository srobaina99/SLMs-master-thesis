# Introduce starters vocabulary

```python
# Read the starter vocabulary from file
with open("./starters_vocab.txt", "r") as f:
    starter_vocabulary = f.read().splitlines()
```

```python
filtered_vocabulary = []
for item in starter_vocabulary:
    # Skip single uppercase letters (section headers)
    if len(item) == 1 and item.isupper():
        continue
    # For other entries, take only the first word
    words = item.split()
    if words:
        filtered_vocabulary.append(f" {words[0]}")

filtered_vocabulary
```

```python
# Save the filtered vocabulary to a text file
with open("./filtered_starters_vocab.txt", "w") as f:
    for word in filtered_vocabulary:
        f.write(f"{word.strip()}\n")

print(f"Saved {len(filtered_vocabulary)} filtered vocabulary words to filtered_starters_vocab.txt")
```
